```python
import mygene
import pandas as pd

# 1. Read gene names from your .txt file
txt_file_path = "gene_names.txt"  # Replace with your actual file path

with open(txt_file_path, "r") as f:
    gene_list = [line.strip() for line in f if line.strip()]

print(f"Loaded {len(gene_list)} genes from {txt_file_path}.")

# 2. Initialize the MyGene client
mg = mygene.MyGeneInfo()

# 3. Query the database for GO Biological Process (BP) annotations
print("Fetching data from MyGene.info...")
results = mg.querymany(
    gene_list,
    scopes="symbol",
    fields="go.BP",
    species="human",  # Change to your organism if not human
    returnall=False
)

# 4. Parse and club GO IDs and Functions together per gene
parsed_data = []

for entry in results:
    gene_name = entry.get("query")
    go_data = entry.get("go", {})
    
    # Check if Biological Process (BP) data exists
    if "BP" in go_data:
        bp_list = go_data["BP"]
        
        # Ensure bp_list is always treated as a list of dictionaries
        if isinstance(bp_list, dict):
            bp_list = [bp_list]
            
        # Extract IDs and Terms
        go_ids = [term.get("id") for term in bp_list if term.get("id")]
        functions = [term.get("term") for term in bp_list if term.get("term")]
        
        # Club them into comma-separated strings
        combined_ids = ", ".join(go_ids)
        combined_functions = ", ".join(functions)
        
    else:
        combined_ids = "None Found"
        combined_functions = "None Found"
        
    parsed_data.append({
        "Gene Name": gene_name,
        "GO IDs": combined_ids,
        "Biological Functions": combined_functions
    })

# 5. Save the combined results to a CSV file
df = pd.DataFrame(parsed_data)
output_file = "combined_gene_go_data.csv"
df.to_csv(output_file, index=False)

print(f"\nSuccess! Results saved to '{output_file}'")
print(df.head())
```

```python
panther_df = pd.read_csv(
    "pantherGeneList_2.txt",
    sep="\t",
    header=None
)
print(panther_df.head().to_string())
panther_df.columns = [
    "gene_identifier",
    "gene_identifier_2",
    "gene_description",
    "organism",
    "go_annotations",
    "pathways"
]
panther_df.head()
```

```python
panther_df["gene_name"] = (
    panther_df["gene_description"]
    .str.split(";")
    .str[1]
    .str.strip()
    .str.upper()
)
```

```python
go_df = panther_df[
    ["gene_name", "go_annotations"]
].copy()
```

```python
go_df["go_annotation"] = (
    go_df["go_annotations"]
    .str.split(";")
)

go_df = go_df.explode("go_annotation")
```

```python
go_df["go_annotation"] = (
    go_df["go_annotations"]
    .str.split(";")
)

go_df = go_df.explode("go_annotation")
```

```python
go_df["go_annotation"] = (
    go_df["go_annotation"]
    .str.strip()
)
```

```python
go_df = go_df[
    go_df["go_annotation"].notna()
    & go_df["go_annotation"].ne("")
]
```

```python
go_df["go_id"] = (
    go_df["go_annotation"]
    .str.extract(r"(GO:\d+)", expand=False)
)
```

```python
go_df["biological_function"] = (
    go_df["go_annotation"]
    .str.replace(r"\s*\(GO:\d+\)", "", regex=True)
    .str.strip()
)
```
